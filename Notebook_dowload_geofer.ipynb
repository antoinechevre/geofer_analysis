```python
import os
import requests

# Ressources du jeu de donnees Geofer (data.gouv.fr)
# https://www.data.gouv.fr/datasets/geofer-donnees-de-potentiel-territorial-des-gares-ferroviaires
GEOFER_RESOURCES = {
    "geofer_gares.csv": "https://www.data.gouv.fr/api/1/datasets/r/20799463-df66-4148-a35d-3f9366f3bc66",
    "iso_10min_voiture.geojson": "https://www.data.gouv.fr/api/1/datasets/r/f8eeb70a-0244-413d-9ccd-b1a88c176090",
    "iso_10min_velo.geojson": "https://www.data.gouv.fr/api/1/datasets/r/21f99db3-eaa1-4f81-9958-54043ed11b0a",
    "iso_15min_pieton.geojson": "https://www.data.gouv.fr/api/1/datasets/r/934c20b2-2df9-4c07-844a-b73f90573cdc",
}

DATA_DIR = "Data_geofer"
os.makedirs(DATA_DIR, exist_ok=True)

for filename, url in GEOFER_RESOURCES.items():
    dest_path = os.path.join(DATA_DIR, filename)
    if os.path.exists(dest_path):
        print(f"Deja present, on passe : {dest_path}")
        continue
    print(f"Telechargement de {filename}...")
    response = requests.get(url, allow_redirects=True)
    response.raise_for_status()
    with open(dest_path, "wb") as f:
        f.write(response.content)
    print(f"  -> enregistre dans {dest_path} ({len(response.content) / 1e6:.1f} Mo)")
```
